```python
import pandas as pd

df = pd.read_csv("../data/sales.csv")
```

```python
print(df.head())
```

```python
print(df.columns.tolist())
```

```python
product_sales = df.groupby("Product")["Sales"].sum().sort_values(ascending=False)

print(product_sales)
```

```python
region_product_sales = (
    df.groupby(["Region", "Product"])["Sales"]
      .sum()
      .reset_index()
)

region_product_sales
```

```python
best_product_by_region = (
    region_product_sales.loc[
        region_product_sales.groupby("Region")["Sales"].idxmax()
    ]
)

best_product_by_region
```

```python
total_sales = df["Sales"].sum()
average_sales = df["Sales"].mean()
number_of_orders = df.shape[0]
highest_sale = df["Sales"].max()
lowest_sale = df["Sales"].min()

print("Total Sales:", total_sales)
print("Average Sale:", average_sales)
print("Number of Orders:", number_of_orders)
print("Highest Sale:", highest_sale)
print("Lowest Sale:", lowest_sale)
```

```python
import matplotlib.pyplot as plt
product_sales = df.groupby("Product")["Sales"].sum().sort_values(ascending=False)

plt.figure(figsize=(10, 6))
product_sales.plot(kind="bar")

plt.title("Sales by Product")
plt.xlabel("Product")
plt.ylabel("Total Sales")
plt.xticks(rotation=45)
plt.tight_layout()

plt.savefig("../visualizations/sales_by_product.png", dpi=300, bbox_inches="tight")

plt.show()
```

```python
region_sales = df.groupby("Region")["Sales"].sum().sort_values(ascending=False)

plt.figure(figsize=(8, 5))
region_sales.plot(kind="bar")

plt.title("Sales by Region")
plt.xlabel("Region")
plt.ylabel("Total Sales")
plt.xticks(rotation=0)
plt.tight_layout()

plt.savefig("../visualizations/sales_by_region.png", dpi=300, bbox_inches="tight")

plt.show()
```

```python
category_sales = df.groupby("Category")["Sales"].sum().sort_values(ascending=False)

plt.figure(figsize=(8, 5))
category_sales.plot(kind="bar")

plt.title("Sales by Category")
plt.xlabel("Category")
plt.ylabel("Total Sales")
plt.xticks(rotation=0)
plt.tight_layout()

plt.savefig(
    "../visualizations/sales_by_category.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```

```python
low_products = (
    df.groupby("Product")["Sales"]
    .sum()
    .sort_values()
    .head(5)
)

plt.figure(figsize=(8, 5))
low_products.plot(kind="bar")

plt.title("Lowest-Performing Products")
plt.xlabel("Product")
plt.ylabel("Total Sales")
plt.xticks(rotation=45)
plt.tight_layout()

plt.savefig(
    "../visualizations/lowest_performing_products.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```

```python
print("===== BUSINESS INSIGHTS =====")

print("1. Total Sales:", total_sales)

print("2. Best-Selling Product:", product_sales.idxmax())

print("3. Best-Performing Region:", region_sales.idxmax())

print("4. Best-Performing Category:", category_sales.idxmax())

print("5. Lowest-Performing Product:", low_products.idxmin())

print("6. Average Sale:", average_sales)
```
